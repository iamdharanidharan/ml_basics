```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("/content/medical_insurance.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder=LabelEncoder()
```

```python
data['sex'] = label_encoder.fit_transform(data['sex'])
data['smoker'] = label_encoder.fit_transform(data['smoker'])
data['region'] = label_encoder.fit_transform(data['region'])
```

```python
data.describe()
```

```python
data.head()
```

```python
data.groupby('age')['charges'].describe().mean()
```

```python
x = data[['age','sex','bmi','children','smoker','region']]
y = data[['charges']]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x, y, train_size=0.75)
```

```python
from sklearn.linear_model import LinearRegression
regression_model = LinearRegression()
```

```python
regression_model.fit(x_train,y_train)
```

```python
predictions = regression_model.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r_square_score = r2_score(y_test, predictions)
from sklearn.metrics import mean_squared_error
mse_score = mean_squared_error(y_test, predictions)
from sklearn.metrics import root_mean_squared_error
rmse_score = root_mean_squared_error(y_test, predictions)
from sklearn.metrics import mean_absolute_error
mae_score = mean_absolute_error(y_test, predictions)
```

```python
print('R2 = ' , r_square_score)
print('MSE = ', mse_score)
print('RMSE = ', rmse_score)
print('MASE = ', mae_score)
```

```python
pred_sample = predictions[:5].flatten()
actual_sample = y.values.flatten()[:5] if hasattr(y, 'values') else y[:5]

print(f"{'Predicted Charges':<20} | {'Actual Charges'}")
print("-" * 40)
for pred, actual in zip(pred_sample, actual_sample):
    print(pred, ' | ', actual)
```
